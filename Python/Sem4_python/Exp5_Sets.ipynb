Lambda Functions

Lambda functions are small anonymous functions.They dont have name. Created using
the lambda keyword for short, simple operations.
1. No name unless you assign it to a variable
2. Only one SINGLE LINE expression
3. Implicit return of that expression

```python
func = lambda x: "Positive" if x > 0 else "Negative" if x < 0 else "Zero"
print(func(5))
print(func(-3))
print(func(0))
```

Using with List Comprehension

Lambda functions can be combined with list comprehensions to apply the same
operation to multiple values

```python
func = [lambda arg=x: arg * 10 for x in range(1, 5)]
for i in func:
    print(i())
```

Using for Returning Multiple Results

A lambda can contain only one expression, it can still return multiple results by
combining them into a tuple.

```python
calc = lambda x, y: (x + y, x * y)
res = calc(3, 4)
print(res)
```

Sets

A set in Python is an unordered collection of unique elements. It is defined using curly braces {} or the set() function.
1. No duplicate elements. If try to insert the same item again, it overwrites previous
one.
2. An unordered collection. When we access all items, they are accessed without
any specific order and we cannot access items using indexes as we do in lists.
3. Only add and remove can be used.

s = {1, 2, 3}

s = set([1, 2, 3])

s = set()

```python
s = {10, 50, 20}
print(s)
print(type(s))
# typecasting list to set
s = set(["a", "b", "c"])
print(s)
# a set cannot have duplicate values
s = {"Geeks", "for", "Geeks"}
print(s)
# values of a set cannot be changed
#s[1] = "Hello"
#print(s)
```

Methods for Sets
1. Adding elements to Sets: add() function is used to insert new elements into a
set. It automatically ignores duplicates.

```python
s = {"a", "b", "c"}
s.add("d")
s.remove("c")
print(s)
```

Union of Sets: union() function combines two sets and returns a new set with all unique
elements.

```python
a={"x","y"}
b={"y","x"}
u=a.union(b)
print(a|b)
print(u)
```

Intersection of Sets: intersection() function returns a new set containing elements that
are common to both sets.

```python
a={1,2,3}
b={2,3,4}
i=a.intersection(b)
print(a&b)
print(i)
```

Difference of Sets: difference() function returns a set containing elements that are in the first set but not in the second.

```python
a={1,2,3}
b={2,3,4}
d=a.difference(b)
print(a-b)
print(a^b)#symmetric difference
print(d)
```

Clearing a Set

clear() function removes all elements from a set, leaving it empty.

```python
s = {1, 2, 3}
s.clear()
print(s)
```

A is subset of B if all elements of A are in B.

```python
A = {1, 2}
B = {1, 2, 3, 4}
print(A.issubset(B))
```

B is superset of A

```python
print(B.issuperset(A))
print(3 in {1, 2, 3})
for item in {1, 2, 3}:
    print(item)
```
