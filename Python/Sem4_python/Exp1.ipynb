Python Comparison Operators

• == Equal

• != Not Equal

• <> Two Operands are not equal

• > Greater Than

• < Less Than

• >= Greater Than or Equal to

• <= Less Than or Equal to

```python
x = 20
y = 25
print('x == y is',x==y)
print('x != y is',x!=y)
print('x > y  is',x>y)
print('x < y  is',x<y)
print('x >= y is',x>=y)
print('x <= y is',x<=y)
```

Python Logical Operators

• And - Logical AND

• Or - Logical OR

• Not - Logical NOT

```python
x = 10
y = 20
print('x and y is',x and y)
print('x or y is',x or y)
print('not x is',not x)
```

Python Bitwise operators

• & Binary AND

• | Binary OR

• ^ Binary XOR

• ~ Binary Ones Complement

• << Binary Left Shift

• >> Binary Right Shift

```python
a = 10
b = 4
print(a & b)
print(a | b)
print(a ^ b)
print(~a)
print(a << 2)
print(a >> 2)
```

Python Membership Operators
“In” and “Not In” are the two membership operators available in Python.

```python
a  = 10
b  =  7
list = [1,2,3,4,5,6,7,8,9]
if (a in list):
    print(a," is presented in the list")
else :
    print(a, "is not presented in the list")
if (b in list):
    print(b, "is presented in the list")
else:
    print(b, "is not presented in the list")
```

```python
x = 23
y = 20
l = [10, 15,20,  30, 35, 40,45, 50]
if x not in l:
    print("x is not available in the list")
else:
    print("x is available in the  given list")
if y not in l:
    print("x is available in the  given list")
else :
    print("y is available in the given list")
```

Python Identity Operators

We have two identity operators in Python namely is and is not. These are also called as
special operators and are used to find if two identical values (or variables) are located
on the same part of the memory or not. If two variables are equal, that does not mean
they are identical. We have two different operators.

```python
x = 6
if type(x) is int:
    print ("true")
else:
    print ("false")
x = 4.5
if (type (x) is not int):
    print ("true")
else:
    print ("false")
```
