```python
import pandas as pd

import matplotlib.pyplot as plt

# Create dataset using dictionary

data = {

"Name": ["Aanya", "Riya", "Kabir", "Arjun"],

"Maths": [85, 90, 78, 88],

"Science": [92, 87, 80, 85],

"English": [88, 91, 75, 82]

}

# Create DataFrame

df = pd.DataFrame(data)

# Display full DataFrame

print("Full DataFrame:")

print(df)

# Display selected columns

print("\nSelected Columns (Name and Maths):")

print(df[["Name", "Maths"]])

# Plot graph

df.plot(x="Name", y="Maths", kind="bar")

plt.title("Maths Marks of Students")

plt.show()
```
